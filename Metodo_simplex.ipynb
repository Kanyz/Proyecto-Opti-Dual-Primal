```python
# ============================================================
# RESOLUCIÓN DEL PROBLEMA PRIMAL Y DUAL
# Programación Lineal con scipy.optimize.linprog
# ============================================================

from scipy.optimize import linprog
import numpy as np

# ============================================================
# PROBLEMA PRIMAL
# ============================================================

# Función objetivo:
# Min Z = 32x1 + 24x2 + 38x3 + 18x4

c = [32, 24, 38, 18]

# ============================================================
# MATRIZ DE RESTRICCIONES
# ============================================================

# Restricciones originales:
#
# 8x1 + 4x2 + 6x3 + 3x4 >= 150
# 5x1 + 3x2 + 4x3 +10x4 >= 120
# 9x1 + 5x2 + 7x3 + 2x4 >= 180
# 4x1 +10x2 + 5x3 + 6x4 >= 130
#
# linprog trabaja con:
# Ax <= b
#
# entonces multiplicamos todo por -1

A = [
    [-8, -4, -6, -3],
    [-5, -3, -4, -10],
    [-9, -5, -7, -2],
    [-4, -10, -5, -6]
]

b = [-150, -120, -180, -130]

# ============================================================
# RESTRICCIONES DE NO NEGATIVIDAD
# ============================================================

# xi >= 0

bounds = [
    (0, None),
    (0, None),
    (0, None),
    (0, None)
]

# ============================================================
# RESOLVER EL PROBLEMA PRIMAL
# ============================================================

resultado_primal = linprog(
    c=c,
    A_ub=A,
    b_ub=b,
    bounds=bounds,
    method='highs'
)

# ============================================================
# MOSTRAR RESULTADOS PRIMAL
# ============================================================

print("======== PROBLEMA PRIMAL ========")

print("\nVariables óptimas:")

print("x1 =", resultado_primal.x[0])
print("x2 =", resultado_primal.x[1])
print("x3 =", resultado_primal.x[2])
print("x4 =", resultado_primal.x[3])

print("\nCosto mínimo Z =")
print(resultado_primal.fun)

# ============================================================
# PROBLEMA DUAL
# ============================================================

# El dual es:
#
# Max W = 150y1 + 120y2 + 180y3 + 130y4
#
# scipy minimiza, entonces:
#
# Min -W

c_dual = [-150, -120, -180, -130]

# ============================================================
# RESTRICCIONES DEL DUAL
# ============================================================

# 8y1 + 5y2 + 9y3 + 4y4 <= 32
# 4y1 + 3y2 + 5y3 +10y4 <= 24
# 6y1 + 4y2 + 7y3 + 5y4 <= 38
# 3y1 +10y2 + 2y3 + 6y4 <= 18

A_dual = [
    [8, 5, 9, 4],
    [4, 3, 5, 10],
    [6, 4, 7, 5],
    [3, 10, 2, 6]
]

b_dual = [32, 24, 38, 18]

# ============================================================
# RESTRICCIONES DE NO NEGATIVIDAD DUAL
# ============================================================

bounds_dual = [
    (0, None),
    (0, None),
    (0, None),
    (0, None)
]

# ============================================================
# RESOLVER EL DUAL
# ============================================================

resultado_dual = linprog(
    c=c_dual,
    A_ub=A_dual,
    b_ub=b_dual,
    bounds=bounds_dual,
    method='highs'
)

# ============================================================
# MOSTRAR RESULTADOS DUAL
# ============================================================

print("\n======== PROBLEMA DUAL ========")

print("\nVariables duales óptimas:")

print("y1 =", resultado_dual.x[0])
print("y2 =", resultado_dual.x[1])
print("y3 =", resultado_dual.x[2])
print("y4 =", resultado_dual.x[3])

print("\nValor óptimo dual W =")

# recordar cambiar signo porque minimizamos -W
print(-resultado_dual.fun)

# ============================================================
# VERIFICAR DUALIDAD FUERTE
# ============================================================

print("\n======== VERIFICACIÓN ========")

print("Valor primal Z =", resultado_primal.fun)
print("Valor dual W =", -resultado_dual.fun)

# Si ambos coinciden:
# Z = W
# se cumple dualidad fuerte
```
